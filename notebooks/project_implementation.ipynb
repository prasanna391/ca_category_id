```python
from pathlib import Path
import joblib
import pandas as pd
```

```python
BASE_DIR = Path.cwd().parent

MODEL_PATH = BASE_DIR / "models" / "youtube_category_assignable_model.pkl"

print("Project folder:", BASE_DIR)
print("Model path:", MODEL_PATH)
```

```python
model = joblib.load(MODEL_PATH)

print("Model loaded successfully!")
print(model)
```

```python
print("Model type:")
print(type(model))

print("\nModel steps:")
print(model.named_steps)
```

```python
sample_text = "latest smartphone technology review"

prediction = model.predict([sample_text])

print("Input:", sample_text)
print("Predicted category:", prediction[0])
```

```python
samples = [
    "latest smartphone technology review",
    "football match highlights",
    "new cooking recipe",
    "funny animal video",
    "new music performance"
]

predictions = model.predict(samples)

for text, prediction in zip(samples, predictions):
    print("Title:", text)
    print("Category:", prediction)
    print("-" * 50)
```

```python
text = input("Enter a YouTube title or description: ")

if text.strip():
    prediction = model.predict([text])[0]
    print("Predicted category:", prediction)
else:
    print("Please enter some text.")
```
